```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display, HTML
```

```python
# Define the file path relative to project root
file_path = Path("../data/firm_explanations.json")
```

```python
# Read the JSON file
with file_path.open('r') as file:
    firm_explanations = json.load(file)

# Remove features that start with 'Standardized_country'
for company in firm_explanations:
    firm_explanations[company]['top_features'] = [
        feature for feature in firm_explanations[company]['top_features']
        if not feature['feature'].startswith('Standardized_country')
    ]

# Print a sample to verify
print(firm_explanations)
```

```python
# Read the Excel output file
output_df = pd.read_excel('../data/output.xlsx')
output_df = output_df[output_df['Standardized country'] != 'Russia']
output_df.rename(columns={'Standardized country': 'country', 'Company name Latin alphabet': 'company_name'}, inplace=True)

print("Sample of the output DataFrame:")
display(HTML(output_df.head().style.to_html()))
```

```python
# Create a DataFrame with company_name and predicted_proba
data = [{'company_name': company, 'predicted_proba': prob['predicted_proba']} for company, prob in firm_explanations.items()]
df1 = pd.DataFrame(data)

print("Predictions DataFrame:")
display(HTML(df1.head(10).style.to_html()))
```

```python
# Join the first two datasets
df1 = pd.merge(df1, output_df, on='company_name', how='inner')
print("Merged DataFrame Sample:")
display(HTML(df1.head().style.to_html()))
```

```python
# Create a list to store all feature data
feature_data = []

# Iterate through each company and their features
for company_name, data in firm_explanations.items():
    # Get features and their values
    features = data['top_features']
    
    # Sort features by absolute SHAP value to assign ranks
    sorted_features = sorted(features, key=lambda x: abs(x['shap_value']), reverse=True)
    
    # Add each feature with its rank
    for rank, feature in enumerate(sorted_features, 1):
        feature_data.append({
            'company_name': company_name,
            'feature': feature['feature'],
            'feature_value': feature['feature_value'],
            'shap_value': feature['shap_value'],
            'rank': rank
        })

# Create DataFrame
df2 = pd.DataFrame(feature_data)

print("Sample of the feature DataFrame:")
display(HTML(df2.head(10).to_html()))
```

```python
print(df2['feature'].unique())
```

```python
# Function to clean up feature names
def clean_feature_name(feature_name):
    # Add parentheses for "Last avail yr"
    feature_name = feature_name.replace('Last_avail__yr', '(Last avail. yr)')
    # Remove underscores and "EUR"
    feature_name = feature_name.replace('_', ' ').replace('EUR', '').strip()
    # Clean up P/L
    feature_name = feature_name.replace('P L', 'P/L')
    return feature_name

# Apply the function to the feature column
df2['feature'] = df2['feature'].apply(clean_feature_name)

print("Cleaned feature names:")
display(HTML(df2.head(10).to_html()))
```

```python
df2.to_csv('../data/shapley_features.tsv', sep='\t', index=False)
```
