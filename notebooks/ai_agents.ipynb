```python
from dotenv import load_dotenv
import os
import autogen
from autogen import AssistantAgent, UserProxyAgent

load_dotenv()
```

```python
llm_config = { "config_list": [{ "model": "gpt-4", "api_key": os.environ.get("OPENAI_API_KEY") }] }
```

```python
company_name = "FISHBRAIN AB"
country = "Sweden"
```

```python
user_proxy = autogen.UserProxyAgent(
    name="User_proxy",
    system_message="You are a human admin.",
    code_execution_config={
        "last_n_messages": 2,
        "work_dir": "groupchat",
        "use_docker": False,
    }, 
    human_input_mode="NEVER",
)
```

```python
company_researcher = autogen.AssistantAgent(
    name="company_researcher",
    system_message="You are a researcher of companies for a credit risk team in a multinational EU bank. You are responsible for researching the financial information of the company given by your manager. Ensure that you cite and provide links to the articles you use.",
    llm_config=llm_config,
)
```

```python
sector_researcher = autogen.AssistantAgent(
    name="sector_researcher",
    system_message="You are a researcher of sectors and industries in a multinational EU bank.",
    llm_config=llm_config,
)
```

```python
credit_risk_manager = autogen.AssistantAgent(
    name="credit_risk_manager",
    system_message="You are a credit risk manager in a multinational EU bank.",
    llm_config=llm_config,
)
```

```python
tasks = [
    f"""Research any publicly available financial information of {company_name} based in {country} to help with credit risk profiling. Ensure that you cite and provide links to the articles you use.""",
    f"""Research the industry or sector of {company_name} based in {country} to help with credit risk profiling. Ensure that you cite and provide links to the articles you use.""",
    f"""Based on the research of your team, evaluate whether it is appropriate to invest in {company_name} or not. Assume that you do not have their financial information yet, and you are basing only on the context provided.
    If you decide to invest, provide a summary of the research and the reasons for your decision.
    If you decide not to invest, provide a summary of the research and the reasons for your decision.
    Ensure that you cite and provide links to the articles that your researchers used."""
]

```

```python
chat_results = autogen.initiate_chats(
    [
        {
            "sender": user_proxy,
            "recipient": company_researcher,
            "message": tasks[0],
            "clear_history": True,
            "max_turns": 1,
            "summary_method": "last_msg",
        },
        {
            "sender": user_proxy,
            "recipient": sector_researcher,
            "message": tasks[1],
            "max_turns": 1, 
            "summary_method": "last_msg",
        },
        {
            "sender": user_proxy,
            "recipient": credit_risk_manager,
            "message": tasks[2],
            "max_turns": 1, 
            "summary_method": "reflection_with_llm"
        },
    ]
)
```
